# U.S. Medical Insurance Costs

```python
# But first find out how many regions are out there.
```

```python
import csv


def calculate_average(data, count):
    return data/count

def calculate_percentage(data, count):
    return data/count * 100


region_values = set()

with open('insurance.csv', newline='') as insurance_csv:
    reader = csv.DictReader(insurance_csv)
    for row in reader:
        region_values.add(row['region'])

print(list(region_values))
```

```python
# Print the total number of people per region.
```

```python
regions = list(region_values)
northwest_count = 0
northeast_count = 0
southwest_count = 0
southeast_count = 0



with open('insurance.csv', newline='') as insurance_csv:
    reader = csv.DictReader(insurance_csv)
    for row in reader:
        match row['region']:
            case 'northwest':
                northwest_count += 1
            case 'northeast':
                northeast_count += 1
            case 'southwest':
                southwest_count += 1
            case 'southeast':
                southeast_count += 1
            case _:
                print('No data found.')
            

print("Number of contractors per region:")
print(f"1.Northwest: {northwest_count};\n2.Northeast: {northeast_count};\n3.Southwest: {southwest_count};\n4.Southeast: {southeast_count};")
print(f"Total: {northwest_count + northeast_count + southwest_count + southeast_count}")
```

```python
# Calculate the averages charges per region
```

```python
total_charges_northwest = 0
total_charges_northeast = 0
total_charges_southwest = 0
total_charges_southeast = 0

with open('insurance.csv') as insurance_csv:
    reader = csv.DictReader(insurance_csv)
    for row in reader:
        charge = float(row['charges'])
        match row['region']:
            case 'northwest':
                total_charges_northwest += charge
            case 'northeast':
                total_charges_northeast += charge
            case 'southwest':
                total_charges_southwest += charge
            case 'southeast':
                total_charges_southeast += charge
            case _:
                print('No data found')

avg_charge_northwest = calculate_average(total_charges_northwest,northwest_count)
avg_charge_northeast = calculate_average(total_charges_northeast,northeast_count)
avg_charge_southwest = calculate_average(total_charges_southwest,southwest_count)
avg_charge_southeast = calculate_average(total_charges_southeast,southeast_count)



print("Average charge per region:")
print(f"1.Northwest: {avg_charge_northwest:.2f};\n2.Northeast: {avg_charge_northeast:.2f};\n3.Southwest: {avg_charge_southwest:.2f};\n4.Southeast: {avg_charge_southeast:.2f}")
```

```python
#Calculate the average BMI in regions with highest and lowest charges
#This suggests BMI may be one factor related to higher charges in the Southeast, 
#But it does not prove correlation or causation.
```

```python

bmi_southwest = 0
bmi_southeast = 0

with open('insurance.csv') as insurance_csv:
    reader = csv.DictReader(insurance_csv)
    for row in reader:
        region = row['region']
        bmi = float(row['bmi'])
        if region == 'southwest':
            bmi_southwest += bmi
        elif region == 'southeast':
            bmi_southeast += bmi

avg_bmi_southwest = calculate_average(bmi_southwest, southwest_count)
avg_bmi_southeast = calculate_average(bmi_southeast,southeast_count)

print("Correlation between charges and bmi in Southwest(lowest) and Southeast(highest) charges:")
print(f"1.Southwest average bmi: {avg_bmi_southwest:.2f}\n2.Southeast average bmi: {avg_bmi_southeast:.2f}")
```

```python
# Regions with higher average charges also show a higher proportion of smokers.
```

```python
# Correlation between charges and number of smokers in regions with highest and lowest charges
smokers_southwest = 0
smokers_southeast = 0

with open('insurance.csv') as insurance_csv:
    reader = csv.DictReader(insurance_csv)
    for row in reader:
        region = row['region']
        if region == 'southwest' and row['smoker'] == 'yes':
            smokers_southwest += 1
        elif region == 'southeast' and row['smoker'] == 'yes':
            smokers_southeast += 1

smoker_rate_southwest = calculate_percentage(smokers_southwest, southwest_count)
smoker_rate_southeast = calculate_percentage(smokers_southeast, southeast_count)

print("Correlation between charges and number of smokers in Southwest(lowest) and Southeast(highest) charges:")
print(f"1.Southwest active smokers: {smoker_rate_southwest:.2f}%\n2.Southeast active smokers: {smoker_rate_southeast:.2f}%")
    
```

```python
# Correlation between charges and number of males and females in regions with highest and lowest charges
males_southwest = 0
males_southeast = 0

with open('insurance.csv') as insurance_csv:
    reader = csv.DictReader(insurance_csv)
    for row in reader:
        region = row['region']
        if region == 'southwest' and row['sex'] == 'male':
            males_southwest += 1
        elif region == 'southeast' and row['sex'] == 'male':
            males_southeast += 1

male_rate_southwest = calculate_percentage(males_southwest, southwest_count)
male_rate_southeast = calculate_percentage(males_southeast, southeast_count)

print("Correlation between charges and number of males in Southwest(lowest) and Southeast(highest) charges:")
print(f"1.Southwest male population: {male_rate_southwest:.2f}%\n2.Southeast male population: {male_rate_southeast:.2f}%")
```
